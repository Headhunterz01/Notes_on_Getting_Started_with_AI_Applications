### OpenCV中用于目标检测任务的级联分类器模型文件
haarcascade_eye.xml: 用于眼睛检测的级联分类器模型。它可以用于检测图像或视频中的眼睛。

haarcascade_eye_tree_eyeglasses.xml: 用于眼镜和眼睛联合检测的级联分类器模型。它可以检测图像或视频中戴眼镜的眼睛。

haarcascade_fullbody.xml: 用于全身人体检测的级联分类器模型。它可用于在图像或视频中检测整个人体。

haarcascade_lefteye_2splits.xml: 用于检测左眼的级联分类器模型。它可以用于检测图像或视频中的左眼。

haarcascade_licence_plate_rus_16stages.xml: 用于俄罗斯车牌检测的级联分类器模型。它可以用于在图像或视频中检测俄罗斯车牌。

haarcascade_lowerbody.xml: 用于下半身人体检测的级联分类器模型。它可用于检测图像或视频中的下半身（例如腿部）。

haarcascade_profileface.xml: 用于侧脸检测的级联分类器模型。它可以用于检测图像或视频中的侧脸。

haarcascade_righteye_2splits.xml: 用于检测右眼的级联分类器模型。它可以用于检测图像或视频中的右眼。

haarcascade_frontalface_default.xml：这是用于人脸检测的默认级联分类器模型。它是OpenCV中最常用的人脸检测模型之一，用于在图片或视频中检测正脸。

haarcascade_frontalface_alt.xml：这是另一种人脸检测级联分类器模型，相对于默认模型，这个模型在检测率上有所提升，但可能会产生一些误检测。

haarcascade_frontalface_alt2.xml：这是haarcascade_frontalface_alt.xml模型的改进版本，在保持较高检测率的同时，尽可能减少了误检测。

haarcascade_frontalface_alt_tree.xml：这也是一种人脸检测级联分类器模型，采用的是基于树的分类器结构，相比于前面的模型，对于一些复杂情况下的人脸检测可能会更准确，但可能需要更长的计算时间。

haarcascade_frontalcatface.xml：这是用于猫脸检测的级联分类器模型。它专门训练用于检测图像或视频中的猫脸。

haarcascade_frontalcatface_extended.xml：这是扩展版的猫脸检测级联分类器模型，对于一些特殊场景下的猫脸检测可能会更准确。

haarcascade_russian_plate_number.xml：这是用于俄罗斯车牌号码检测的级联分类器模型。它专门用于在图像或视频中检测俄罗斯车牌号码。

haarcascade_smile.xml：这是用于微笑检测的级联分类器模型。它可以用于在图像或视频中检测人脸上的微笑区域。

haarcascade_upperbody.xml：这是用于上半身人体检测的级联分类器模型。它可用于在图像或视频中检测上半身（例如头部和肩部）。

haarcascade_mcs_mouth.xml和haarcascade_mcs_nose.xml是后来找到放进去的，其余都是下载OpenCV之后自带的

```python
# 2.2 眼睛检测
import cv2
import matplotlib.pyplot as plt
imgfile="imgs/face.jpg"
img=cv2.imread(imgfile)
#加载人脸检测模型
harrcascade_path =cv2.data.haarcascades + "haarcascade_frontalface_alt.xml"
face_detector = cv2.CascadeClassifier(harrcascade_path)
eye_path=cv2.data.haarcascades + "haarcascade_eye.xml"
eye_detector = cv2.CascadeClassifier(eye_path)

faces=face_detector.detectMultiScale(img, 1.3,5)    
if len(faces)>0:
    for i,face in enumerate(faces):
        x, y, w, h = face
        roi_face=img[y:y+h,x:x+h]
        eyes=eye_detector.detectMultiScale(roi_face,1.3,5)
        # 在眼睛周围画一个矩形
        for (ex,ey,ew,eh) in eyes:
            cv2.rectangle(roi_face,(ex,ey),(ex+ew,ey+eh),(0,255,255),2)
        cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)
# 显示图像
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.show()
```

```python
# 2.3 OpenCV打开摄像头
import cv2
#打开本地视频
# videofile="imgs/smoke.mp4"
# cap = cv2.VideoCapture(videofile)
#打开本地摄像头
cap = cv2.VideoCapture(0)
#循环读取摄像头的图像
while cap.isOpened():
    #读取摄像头的图像
    ret, frame = cap.read()
    #显示图像
    cv2.imshow('frame', frame)
    #按下q键退出循环
    if cv2.waitKey(1) == ord('q'):
        break
#释放摄像头
cap.release()
#关闭窗口
cv2.destroyAllWindows()
```

```python
# 2.4 opencv 键盘鼠标事件
import cv2
#打开摄像头
cap = cv2.VideoCapture(0)
isgray=False #是否灰度
#定义鼠标回调函数，移动鼠标，显示鼠标位置
def mouse_callback(event, x, y, flags, param):
    if event == cv2.EVENT_MOUSEMOVE:
        cv2.putText(frame, f"({x}, {y})", (x, y), 0, 0.5, (0, 0, 255), 1)
#创建窗口，并设置鼠标回调函数
cv2.namedWindow('frame')
cv2.setMouseCallback('frame', mouse_callback)

while True:
    # 读取摄像头的每一帧
    ret, frame = cap.read()
    # 如果读取帧失败，则退出循环
    if not ret:
        break
    # 按下 'q' 键退出循环
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    # 按下 'g' 键切换灰度模式
    if cv2.waitKey(1) & 0xFF == ord('g'):
        isgray= not isgray
    # 如果是灰度模式，则将每一帧转换为灰度图像
    if isgray:
        frame=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        # 显示每一帧
    cv2.imshow('frame', frame)

# 释放摄像头并关闭窗口
cap.release()
cv2.destroyAllWindows()
```

```python
# 2.5 打开摄像头，实时检测人脸
import cv2
#加载人脸检测模型
harrcascade_path =cv2.data.haarcascades + "haarcascade_frontalface_alt.xml"
face_detector = cv2.CascadeClassifier(harrcascade_path)
#人脸检测，并返回检测结果图像
def facedet(img):
    faces=face_detector.detectMultiScale(img, 1.3,5)    
    if len(faces)>0:
        for i,face in enumerate(faces):
            x, y, w, h = face
            cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)
    return img

cap = cv2.VideoCapture(0)
while cap.isOpened():
    ret,frame = cap.read()
    if ret:
        frame=facedet(frame) ##人脸检测，并返回检测结果图像  
    cv2.imshow("frame",frame)
    if cv2.waitKey(1)==ord("q"):
        break
cap.release()
cv2.destroyAllWindows()
    
```

```python
# 2.6 Opencv录制视频
import cv2
import time

isRecording=False #是否录制
#创建一个录制对象
def CreateVideoWriter():
    #定义视频编码器,mp4v为MPEG-4编码
    fourcc=cv2.VideoWriter_fourcc(*"mp4v")
    filename = time.strftime('%Y%m%d%H%M%S',time.localtime(time.time())) + '.mp4'        
    fourcc=cv2.VideoWriter_fourcc(*"mp4v")
    out=cv2.VideoWriter(filename,fourcc,20,(640,480))
    return out

cap=cv2.VideoCapture(0)#打开摄像头
while cap.isOpened():
    ret,frame=cap.read()    
    img=frame.copy()
    if cv2.waitKey(1)==ord("q"):
        break
    if cv2.waitKey(1)==ord("r"):#开始录制视频       
        isRecording=True      
        out=CreateVideoWriter()  
    if isRecording:        
        out.write(img)        
        #提示用户正在录制视频        
        cv2.putText(frame,"Recording...", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)
    if cv2.waitKey(1)==ord("s"):#停止录制视频
        if isRecording:
            isRecording=False
            out.release
    cv2.imshow("video",frame)
cap.release()
cv2.destroyAllWindows()
```

```python
# 3.2 人脸关键点检测
import cv2
import face_recognition as fr 
import os
import matplotlib.pyplot as plt
import numpy as np 
imgfile="imgs/face.jpg"
img=cv2.imread(imgfile)
print(img.shape)
fp=fr.face_locations(img)
print(fp)
for face in fp:
    #计算人脸的宽高
    h=face[2]-face[0]
    w=face[1]-face[3]
    print(face,h,w)
    #返回人脸图像
    faceimg=img[face[0]:face[2],face[3]:face[1]]
    kp68=fr.face_landmarks(faceimg) #返回关键点
    print(kp68)
    if len(kp68)>0 :
        #绘制关键点
        for key in kp68[0]:
            pts=np.array(kp68[0][key])
            cv2.polylines(faceimg,[pts],False,(255,0,0),1)                                                                                                                   
            for i,xy in enumerate(kp68[0][key]):
                x=xy[0]
                y=xy[1]
                cv2.circle(faceimg,(x,y),2,(0,255,0),1)                                
                cv2.putText(faceimg,str(i),(x,y),cv2.FONT_HERSHEY_SIMPLEX,0.3,(255,0,0),1)
    
    cv2.rectangle(img,(face[3],face[0]),(face[1],face[2]),(0,255,0),2)

# 显示图像
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.show()
```

```python
# 3.3 DLIB
import cv2
import dlib
from matplotlib import pyplot as plt
imgfile="imgs/face.jpg"
img=cv2.imread(imgfile)
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
detector=dlib.get_frontal_face_detector()
#获取特征点模型路径

predictor=dlib.shape_predictor("dlibdat/shape_predictor_68_face_landmarks.dat")
faces=detector(gray)
for face in faces:
    landmarks=predictor(gray,face)
    for n in range(0,68):
        x=landmarks.part(n).x
        y=landmarks.part(n).y
        cv2.circle(img,(x,y),2,(0,255,0),-1)
# 显示图像
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.show()
```

```python
# 4.1 人脸比对
#计算人脸特征向量
import face_recognition as fr
from matplotlib import pyplot as plt 
import cv2

knowface1="imgs/t1.jpg"
unknowface2="imgs/face.jpg"
#读取图像文件返回BGR
#img1=cv2.imread(knowface1)
#img2=cv2.imread(unknowface2)

#读取图像文件返回RGB格式
img1=fr.load_image_file(knowface1)
img2=fr.load_image_file(unknowface2)

img1_embedding = fr.face_encodings(img1)[0]
dstembdeding=fr.face_encodings(img2)
fp=fr.face_locations(img2)
for i,face in enumerate(fp) :
    h=face[2]-face[0]
    w=face[1]-face[3]
    #print(i,h,w)
    img2_embedding =dstembdeding[i]
    #计算欧式距离
    dist = fr.face_distance([img1_embedding], img2_embedding)
    #print(dist)
    #判断是否为同一个人
    #result=fr.compare_faces([img1_embedding], img2_embedding)
    #print(result)
    if dist<0.6:
        dt=1-float(dist[0])
        cv2.putText(img2, f"{i}-Confidence {dt:.1%}:same person", (face[3],face[0]), cv2.FONT_HERSHEY_SIMPLEX,0.4, (255,0, 0), 1)
        print(f"置信度 {dt:.1%}:是同一人")
    else:
        dt=1-float(dist[0])
        cv2.putText(img2, f"{i}-Confidence {dt:.1%}:different person",  (face[3],face[0]), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (0, 255, 0), 1)
        print(f"置信度{dt:.1%}:不是同一个人")
# img1=cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
# img2=cv2.cvtColor(img2,cv2.COLOR_BGR2RGB)
plt.subplot(221)
plt.imshow(img1)
plt.subplot(222)
plt.imshow(img2)
plt.axis('off')
plt.show()
```

```python
# 7.4 行人属性检测
import os
import cv2
import fastdeploy as fd
import numpy as np
import yaml
import paddle.inference as paddle_infer

#行人检测模型
pmodel_file="model/mot_ppyoloe_l_36e_pipeline/inference.pdmodel"
pmodel_params="model/mot_ppyoloe_l_36e_pipeline/inference.pdiparams"
pconfig_file="model/mot_ppyoloe_l_36e_pipeline/inference.yml"

pmodel=fd.vision.detection.PPYOLOE(pmodel_file,pmodel_params,pconfig_file)

#行人属性识别多分类模型
model_file="model/PPHGNet_small_person_attribute_954_infer/inference.pdmodel"
model_params="model/PPHGNet_small_person_attribute_954_infer/inference.pdiparams"
config_file="model/PPHGNet_small_person_attribute_954_infer/infer_cfg.yml"

#读取配置文件,获取标签列表和模型输入尺寸
with open(config_file, 'r') as f:
    config = yaml.safe_load(f)
    label_list = config['label_list']
    
#print(label_list)

def init_predector():
    # 引用 paddle 模型文件
    model_file="model/PPHGNet_small_person_attribute_954_infer/inference.pdmodel"
    params_file="model/PPHGNet_small_person_attribute_954_infer/inference.pdiparams"
    # 配置参数，创建推理对象
    config=paddle_infer.Config(model_file, params_file)
    #配置推理线程数
    config.enable_memory_optim()
    config.set_cpu_math_library_num_threads(4)
    config.enable_mkldnn()

    predictor=paddle_infer.create_predictor(config)
    return predictor

#图像缩放
def resize_short(img, target_size):
    """ resize_short """
    percent = float(target_size) / min(img.shape[0], img.shape[1])
    resized_width = int(round(img.shape[1] * percent))
    resized_height = int(round(img.shape[0] * percent))
    resized = cv2.resize(img, (resized_width, resized_height))
    return resized
#图像裁剪
def crop_image(img, target_size, center):
    """ crop_image """
    height, width = img.shape[:2]
    size = target_size
    if center == True:
        w_start = (width - size) / 2
        h_start = (height - size) / 2
    else:
        w_start = np.random.randint(0, width - size + 1)
        h_start = np.random.randint(0, height - size + 1)
    w_end = w_start + size
    h_end = h_start + size
    img = img[int(h_start):int(h_end), int(w_start):int(w_end), :]
    return img

#图像预处理
def preprocess(img):
    mean = [0.485, 0.456, 0.406]
    std = [0.229, 0.224, 0.225]
    #img = resize_short(img, 256)
    #img = crop_image(img, 256, True)
    # bgr-> rgb && hwc->chw
    img = img[:, :, ::-1].astype('float32').transpose((2, 0, 1)) / 255
    img_mean = np.array(mean).reshape((3, 1, 1))
    img_std = np.array(std).reshape((3, 1, 1))
    img -= img_mean
    img /= img_std
    return img[np.newaxis, :]

def run(predictor, img):
    # copy img data to input tensor
    input_names = predictor.get_input_names()
    for i, name in enumerate(input_names):
        input_tensor = predictor.get_input_handle(name)
        input_tensor.reshape(img[i].shape)
        input_tensor.copy_from_cpu(img[i].copy())

 # do the inference
    predictor.run()

    results = []
    # get out data from output tensor
    output_names = predictor.get_output_names()
    for i, name in enumerate(output_names):
        output_tensor = predictor.get_output_handle(name)
        output_data = output_tensor.copy_to_cpu()
        results.append(output_data)

    return results


imgpath="imgs/77.jpg"
img=cv2.imread(imgpath)
result=pmodel.predict(img)
scores=result.scores
boxes=result.boxes

pred=init_predector()
for box,score in zip(boxes,scores):
    if score>0.5:        
        box=map(lambda x: int(x), box)
        x1, y1, x2, y2 =box
        print(x1,y1,x2,y2)
        cv2.rectangle(img,(x1,y1),(x2,y2),(0,255,0),2)
        pedestrian=img[y1:y2,x1:x2]
        #图像大小缩放成：(h,w)=(256,192)
        pedestrian=cv2.resize(pedestrian, (192, 256))
        print(pedestrian.shape)
        pedestrian=preprocess(pedestrian)
        result=run(pred,[pedestrian])
        #输出每个标签的概率
        for i in range(len(label_list)):
            print(label_list[i],result[0][0][i]>0.5)
        
#可视化
# vis_img= fd.vision.vis_detection(img,result,score_threshold=0.5)
cv2.imshow("result",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 8.1 行人跟踪
import cv2
import os
import fastdeploy as fd
import time

model_file="model/fairmot_hrnetv2_w18_dlafpn_30e_576x320/model.pdmodel"
params_file="model/fairmot_hrnetv2_w18_dlafpn_30e_576x320/model.pdiparams"
config_file="model/fairmot_hrnetv2_w18_dlafpn_30e_576x320/infer_cfg.yml"

opt=fd.RuntimeOption()
opt.use_cpu()
model=fd.vision.tracking.PPTracking(model_file,params_file,config_file,opt)

video_file="imgs/person.mp4"
cap=cv2.VideoCapture(video_file)
#图像预测，返回一个结构体数据
'''
fastdeploy.vision.MOTResult
boxes(list of list(float)): 成员变量，表示单帧画面中检测出来的所有目标框坐标。
boxes是一个list，其每个元素为一个长度为4的list， 表示为一个框，每个框以4个float数值依次表示xmin, ymin, xmax, ymax， 即左上角和右下角坐标
ids(list of list(float)):成员变量，表示单帧画面中所有目标的id，其元素个数与boxes一致
scores(list of float): 成员变量，表示单帧画面检测出来的所有目标置信度
'''
# 预测图片分割结果
frame_id = 0
#fps=int(cap.get(cv2.CAP_PROP_FPS)) #帧率
#ftotal=int(cap.get(cv2.CAP_PROP_FRAME_COUNT))#总帧数
#w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH)) #宽度
#h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT)) #高度
#print(w,h,fps)
while cap.isOpened():
    start_time = time.time()
    frame_id = frame_id+1
    ret, frame = cap.read()
    #frame=cv2.resize(frame,(w//2,h//2))
    if not ret :
        break
    result = model.predict(frame)
    end_time = time.time()
    fps = 1.0/(end_time-start_time)   
    #print(result.boxes,result.ids,result.scores)
    for box,id,scores in zip(result.boxes,result.ids,result.scores):
        if scores>=0.5:
            xmin,ymin,xmax,ymax=box
            cv2.rectangle(frame,(xmin,ymin),(xmax,ymax),(0,255,0),2)
            cv2.putText(frame,str(id),(xmin,ymin-10),cv2.FONT_HERSHEY_SIMPLEX,0.5,(0,255,0),2)
    cv2.putText(frame,"FPS: {:.2f}".format(fps),(10,30),cv2.FONT_HERSHEY_SIMPLEX,1,(0,255,0),2)
    
    cv2.imshow("video", frame)
    if cv2.waitKey(10)==ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# 3.1 图像分类
# 划分图像分类数据集，将文件夹中的图像划分成训练集和验证集，并生成label.txt和train.txt、val.txt文件
import os
import random

# 数据集的根目录，
basedir="LionDance-Other" #
if os.path.exists(os.path.join(basedir, 'label.txt')):
    os.remove(os.path.join(basedir, 'label.txt'))
if os.path.exists(os.path.join(basedir, 'train.txt')):
    os.remove(os.path.join(basedir, 'train.txt'))
if os.path.exists(os.path.join(basedir, 'val.txt')):
    os.remove(os.path.join(basedir, 'val.txt'))

for i,folder in enumerate(os.listdir(basedir)):
    if os.path.isdir(os.path.join(basedir, folder)):
        # Set the path to the directory containing the images
        dir_path = os.path.join(basedir, folder)
        #print(i,folder)
        #写入label.txt,格式：i folder，如果文件存在，清除文件内容      
        with open(os.path.join(basedir,'label.txt'), 'a') as f:
            f.write(f'{i} {folder}\n')
        #遍历文件夹中的所有图像文件，将它们存放到一个列表中，格式为：folder/文件名 i
        image_files = [f'{folder}/{file}' for file in os.listdir(dir_path) if file.lower().endswith('.jpg') or file.lower().endswith('.png') or file.lower().endswith('.jpeg')]
        # 打乱图像文件的顺序
        #print(image_files,len(image_files))
        random.shuffle(image_files)
        # 计算训练集和验证集的大小
        train_size = int(len(image_files) * 0.8)
        # 将图像文件分成训练集和验证集
        train_files = image_files[:train_size]
        val_files = image_files[train_size:]
        # 创建训练集和验证集的文本文件分别train.txt和val.txt
        with open(os.path.join(basedir,'train.txt'), 'a') as f:
            for file in train_files:
                f.write(f'{file} {i}\n')
        with open(os.path.join(basedir,'val.txt'), 'a') as f:
            for file in val_files:
                f.write(f'{file} {i}\n')
        
```

```python
# 3.2 目标检测
import os
import random

#voc数据集划分
train_ratio = 0.8
val_ratio = 0.2
basedir="fruit5_voc"
annodir="Annotations"
imgdir="JPEGImages"
#根据xml文件内容，生成labels.txt
def generate_labels():
    labels = []
    for filename in os.listdir(os.path.join(basedir,annodir)):
        if filename.endswith(".xml"):
            with open(os.path.join(basedir,annodir, filename), "r") as f:
                for line in f:
                    if "<name>" in line:
                        label = line.strip().replace("<name>", "").replace("</name>", "")
                        labels.append(label)
    with open("labels.txt", "w") as f:
        for label in set(labels):
            f.write(label + "\n")

generate_labels() #生成labels.txt

#划分数据集,生成train.txt和val.txt,格式为：图片路径 xml路径
def split_dataset():
    with open("train_list.txt", "w") as f_train, open("val_list.txt", "w") as f_val:
        for filename in os.listdir(os.path.join(basedir,imgdir)):
            if filename.endswith(".jpg") or filename.endswith(".png") or filename.endswith(".jpeg"):
                xml_filename = os.path.splitext(filename)[0] + ".xml"
                if random.random() < train_ratio: # 80%的数据用于训练集
                    f_train.write(os.path.join(imgdir, filename) + " " + os.path.join(annodir, xml_filename) + "\n")
                else:
                    f_val.write(os.path.join(imgdir, filename) + " " + os.path.join(annodir, xml_filename) + "\n")

split_dataset() #划分数据集，生成train.txt和val.txt



```

```python
# 14.3 检测是否张开嘴巴，如果张开嘴巴，则脸部马赛克处理
import cv2
import face_recognition as fr 
import os
import numpy as np 
#检测人脸函数
def mouthdist(img):
    fp=fr.face_locations(img)
    #print(fp)
    for face in fp:
        #计算人脸的宽高
        h=face[2]-face[0]
        w=face[1]-face[3]
        #print(face,h,w)
        #返回人脸图像
        faceimg=img[face[0]:face[2],face[3]:face[1]]
        kp68=fr.face_landmarks(faceimg) #返回关键点
        if len(kp68)>0:
            top_lip=kp68[0]["top_lip"]        
            bottom_lip=kp68[0]["bottom_lip"]      
            #上嘴唇和下嘴唇
            # print(top_lip)
            # print(bottom_lip)
            #计算上嘴唇第8，9，10的特点与下嘴唇第10,9,8点的欧氏距离
            tlip=np.array(top_lip[8:11])
            blip=np.array(bottom_lip[8:11])
            blip=blip[::-1]
            #计算欧氏距离
            dist=np.linalg.norm(tlip-blip,axis=1)
            dd=round(np.mean(dist),0)          
                    
        #绘制关键点        
            for key in kp68[0]:
                pts=np.array(kp68[0][key])
                cv2.polylines(faceimg,[pts],False,(255,0,0),1)                                                                                                                   
                for i,xy in enumerate(kp68[0][key]):
                    x=xy[0]
                    y=xy[1]
                    cv2.circle(faceimg,(x,y),2,(0,255,0),1)                                
                    cv2.putText(faceimg,str(i),(x,y),cv2.FONT_HERSHEY_SIMPLEX,0.3,(255,0,0),1)
            
            cv2.rectangle(img,(face[3],face[0]),(face[1],face[2]),(0,255,0),2)

            return dd,face
            
    return None,None

#马赛克
def mosaic(img,h,w,sz=10):    
    for i in range(0,h,sz):
        for j in range(0,w,sz):
            sub=img[i:i+sz,j:j+sz]
            m=cv2.mean(sub)[:3]
            img[i:i+sz,j:j+sz]=m
    return img

cap=cv2.VideoCapture(0)
while cap.isOpened():
    ret,frame=cap.read()
    if ret:
        dist,face=mouthdist(frame) #返回嘴巴距离和人脸位置
        if face is not None:
            h=face[2]-face[0]
            w=face[1]-face[3]           
            faceimg=frame[face[0]:face[2],face[3]:face[1]]
            if dist>15:
                frame[face[0]:face[2],face[3]:face[1]]=mosaic(faceimg,h,w)        
        cv2.putText(frame,"dist:{}".format(dist),(10,30),cv2.FONT_HERSHEY_SIMPLEX,1,(0,255,0),2)
        cv2.imshow("frame",frame)
        if cv2.waitKey(1)==ord("q"):
            break
        if cv2.getWindowProperty("headpose",cv2.WND_PROP_VISIBLE) < 1:
            break

cap.release()
cv2.destroyAllWindows()
```

```python
# 14.4 实时人头检测+头部姿态估计 CPU运行速度很慢
#euler_angles(list of float): 成员变量，表示单张人脸图片预测的欧拉角，存放的顺序是(yaw, pitch, roll)， 
#yaw 代表水平转角，pitch 代表垂直角，roll 代表翻滚角，值域都为 [-90,+90]度

import cv2
import fastdeploy as fd
import os

#头部检测
modeldir="model/faster_rcnn_r50_person"
model_file=os.path.join(modeldir,"inference.pdmodel")
param_file=os.path.join(modeldir,"inference.pdiparams")
config_file=os.path.join(modeldir,"inference.yml")
detector=fd.vision.detection.FasterRCNN(model_file,param_file,config_file)


# opt=fd.RuntimeOption()
# opt.use_gpu()
# opt1use_paddle_backend()

#头部姿态估计
fsanet_model_file="model/FSAnet/fsanet-var.onnx"
fsanet=fd.vision.headpose.FSANet(model_file=fsanet_model_file,
                                 params_file="",
                                 runtime_option=None,
                                 model_format=fd.ModelFormat.ONNX)

#打开摄像头
cap=cv2.VideoCapture(0)
while cap.isOpened():
    ret,frame=cap.read()
    if not ret:
        break
    result=detector.predict(frame)
    scores=result.scores
    boxes=result.boxes
    for score,box in zip(scores,boxes):
        if score>0.5:
            x,y,w,h=int(box[0]),int(box[1]),int(box[2]-box[0]),int(box[3]-box[1])        
            faceimg=frame[y:y+h,x:x+w]
            #画框
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
            faceimg=cv2.cvtColor(faceimg, cv2.COLOR_BGR2RGB)
            result=fsanet.predict(faceimg)
            vis_im = fd.vision.vis_headpose(faceimg, result)  
            vis_im=cv2.cvtColor(vis_im, cv2.COLOR_RGB2BGR)   
            frame[y:y+h,x:x+w]=vis_im
            #实时显示euler_angles
            euler_angles="yaw: {:.0f},pitch: {:.0f},roll: {:.0f}".format(result.euler_angles[0], result.euler_angles[1], result.euler_angles[2])
            cv2.putText(frame, euler_angles, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.imshow("headpose", frame)

    if cv2.waitKey(1)==ord("q"):
        break
    if cv2.getWindowProperty("headpose",cv2.WND_PROP_VISIBLE) < 1:
        break
cap.release()
cv2.destroyAllWindows()
```

```python
# 14.8 检测眼睛开闭程序
import cv2
import face_recognition as fr 
import os
import numpy as np 
import time
#检测人脸函数
def eyedist(img):
    fp=fr.face_locations(img)
    #print(fp)
    for face in fp:
        #计算人脸的宽高
        h=face[2]-face[0]
        w=face[1]-face[3]
        #print(face,h,w)
        #返回人脸图像
        faceimg=img[face[0]:face[2],face[3]:face[1]]
        kp68=fr.face_landmarks(faceimg) #返回关键点
        if len(kp68)>0:
            #返回左眼和右眼的坐标
            left_eye=kp68[0]["left_eye"]
            right_eye=kp68[0]["right_eye"]
            #print(left_eye,right_eye)

            #计算左右眼张开的距离
            leyeup=np.array(left_eye[1:3])
            leyedown=np.array(left_eye[4:6])
            leyedown=leyedown[::-1]
            reyeup=np.array(right_eye[1:3])
            reyedown=np.array(right_eye[4:6])
            reyedown=reyedown[::-1]
            ddl=np.linalg.norm(leyeup-leyedown,axis=1)
            ddr=np.linalg.norm(reyeup-reyedown,axis=1)
            dd=round((np.mean(ddl)+np.mean(ddr))/2,0)
            #print(dd)
            
        #绘制关键点        
            for key in kp68[0]:
                pts=np.array(kp68[0][key])
                cv2.polylines(faceimg,[pts],False,(255,0,0),1)                                                                                                                   
                for i,xy in enumerate(kp68[0][key]):
                    x=xy[0]
                    y=xy[1]
                    cv2.circle(faceimg,(x,y),2,(0,255,0),1)                                
                    cv2.putText(faceimg,str(i),(x,y),cv2.FONT_HERSHEY_SIMPLEX,0.3,(255,0,0),1)
            
            cv2.rectangle(img,(face[3],face[0]),(face[1],face[2]),(0,255,0),2)

            return dd,face
            
    return None,None


cap=cv2.VideoCapture(0)
while cap.isOpened():
    ret,frame=cap.read()
    if ret:
        framecopy=frame.copy()
        dist,face=eyedist(frame) #返回嘴巴距离和人脸位置
        if face is not None:
            h=face[2]-face[0]
            w=face[1]-face[3]           
            faceimg=frame[face[0]:face[2],face[3]:face[1]]
            if dist<11:
                #以时间戳命名保存图片
                fname=time.strftime("%Y%m%d%H%M%S",time.localtime())+".jpg"
                cv2.imwrite("output/"+fname,framecopy)
                
                   
        cv2.putText(frame,"dist:{}".format(dist),(10,30),cv2.FONT_HERSHEY_SIMPLEX,1,(0,255,0),2)
        cv2.imshow("frame",frame)
        if cv2.waitKey(1)==ord("q"):
            break
        if cv2.getWindowProperty("frame",cv2.WND_PROP_VISIBLE)<=0:
            break

cap.release()
cv2.destroyAllWindows()
```

```python
# 14.9 模拟人脸录入
import cv2
import face_recognition as fr
import os
import time

facedir="faces"
facevlst=[]
#读取facedir人脸库,将人脸编码保存到编码列表facesvlst中
def load_faces():
    for filename in os.listdir(facedir):
        img_path=os.path.join(facedir,filename)
        img=fr.load_image_file(img_path)
        face_encoding=fr.face_encodings(img)[0]
        facevlst.append(face_encoding)
    return facevlst
facevlst=load_faces()
#print("人脸编码加载完成",len(facevlst),"张人脸编码")

#定义图像与人脸库比较函数
def compare_faces(face_encoding,unknownimg):
    #如果图库为空，则返回False
    if len(facevlst)==0:
        return False
    #检测unknownimg中的人脸
    faces=fr.face_locations(unknownimg)
    if len(faces)==0:
        return False
    face_encoding=fr.face_encodings(unknownimg)[0]
    for vlst in facevlst:    
        distances=fr.face_distance([vlst],face_encoding)
        if distances<0.6:
            return True
    return False

msgint=0 #显示提示不同信息标记 0:无提示 1:人脸已存在 2:未检测到人脸
#打开摄像头
cap=cv2.VideoCapture(0)
while cap.isOpened():
    #读取摄像头画面
    ret,frame=cap.read()
    if not ret:
        break
   
    #按下p键保存当前人脸
    if cv2.waitKey(1)==ord('p'):
        facevlst=load_faces()
        #人脸检测
        faces=fr.face_locations(frame)
        #如果检测到人脸，则保存当前帧
        if len(faces)>0:    
            is_exist=compare_faces(facevlst,frame)
            if is_exist:
                msgint=1
                print("人脸已存在")
            else:                
                #将当前帧保存为图片,以时间戳命名
                fname=time.strftime("%Y%m%d%H%M%S")+".jpg"
                cv2.imwrite(os.path.join(facedir,fname),frame)
                print("人脸保存成功")
                msgint=0
        else:
            msgint=2
            print("未检测到人脸")
    #显示提示信息
    if msgint==1:
        cv2.putText(frame,"Face exist",(10,60),cv2.FONT_HERSHEY_SIMPLEX,1,(0,255,0),2)
    elif msgint==2:
        cv2.putText(frame,"No face detected",(10,60),cv2.FONT_HERSHEY_SIMPLEX,1,(0,0,255),2)
    elif msgint==0:
        cv2.putText(frame,"Press 'p' to save face",(10,30),cv2.FONT_HERSHEY_SIMPLEX,1,(0,255,0),2)
    cv2.imshow('frame',frame)

    
    if cv2.waitKey(1)==ord('q'):
        break
    if cv2.getWindowProperty('frame',cv2.WND_PROP_VISIBLE) < 1:
        break  
    
     
cap.release()
cv2.destroyAllWindows()
```

```python
# 14.11 图像分类GUI版
import cv2
import fastdeploy as fd
import os 
import time
import threading
from tkinter import filedialog
from tkinter import messagebox
from tkinter import ttk
import tkinter as tk
from PIL import Image, ImageTk, ImageOps

#创建一个主窗口
root = tk.Tk()
root.title("图像分类")
root.geometry("800x600")

#图像分类预测
# 加载模型
model_dir="model/PPLCNetV2_base_ssld" # ImageNet1K
model_file = os.path.join(model_dir, "inference.pdmodel")
params_file = os.path.join(model_dir, "inference.pdiparams")
config_file = os.path.join(model_dir, "inference.yml")
def predict_image(img):
    model = fd.vision.classification.PaddleClasModel(model_file, params_file, config_file)
    result = model.predict(img)
    ids=result.label_ids
    scores=result.scores
    return ids,scores

classnames=[]
with open(os.path.join(model_dir,"cnlabels.txt"), "r", encoding="utf-8") as f:
    lines=f.readlines()
    for line in lines:
        classnames.append(line.replace("\n",""))

# print(classnames)

# 创建一个画布，用于显示图像
cw,ch = 640,480
padx,pady = 80,10
canvas = tk.Canvas(root, width=cw, height=ch)
canvas.grid(row=0, column=0, padx=padx, pady=pady)

# 创建一个标签，用于显示分类结果
label = tk.Label(root, text="", font=("", 20))
label.grid(row=1, column=0,padx=padx, pady=pady)

image=None
imgtk=None

def select_image():
    global image,imgtk
    # 打开文件选择对话框，选择图像文件
    file_path = filedialog.askopenfilename()
    if not file_path:
        return
    # 加载图像
    image = Image.open(file_path)
    #调整图像大小以适应画布
    image = ImageOps.fit(image, (cw, ch), Image.Resampling.LANCZOS)
    imgtk = ImageTk.PhotoImage(image)
    # 在画布上显示图像
    canvas.create_image(0, 0, anchor=tk.NW, image=imgtk)
    # 在标签上显示分类结果
    ids,scores=predict_image(cv2.imread(file_path))
    #print(ids,scores)
    label.config(text=f"{classnames[ids[0]]} {scores[0]:.2f}")

    
# 创建一个按钮，用于选择图像
button = tk.Button(root, text="打开图像", command=select_image)
button.grid(row=2, column=0,padx=padx, pady=pady)
   
root.mainloop()
```

```python
# 15.4 目标检测模型训练
#划分数据集
#paddlex --split_dataset --format VOC --dataset_dir bottle --val_value 0.2 --test_value 0.1
import paddlex as pdx
import os
from paddlex import transforms as T 
#定义训练和验证时的transforms

train_transforms=T.Compose([
    T.RandomResizeByShort(short_sizes=[352,384,416],max_size=448,interp="CUBIC"),
    T.RandomHorizontalFlip(),
    T.Normalize()
])
eval_transforms=T.Compose([
    T.ResizeByShort(short_size=416,max_size=448,interp="CUBIC"),
    T.Normalize()
])

#定义训练和验证用的数据集
train_dataset=pdx.datasets.VOCDetection(
    data_dir="bottle",
    file_list="bottle/train_list.txt",
    label_list="bottle/labels.txt",
    transforms=train_transforms,
    shuffle=True
)
eval_dataset=pdx.datasets.VOCDetection(
    data_dir="bottle",
    file_list="bottle/val_list.txt",
    label_list="bottle/labels.txt",
    transforms=eval_transforms,
    shuffle=False
)

#初始化模型，并进行训练
num_classes=len(train_dataset.labels)
model=pdx.det.FasterRCNN(num_classes=num_classes,backbone="ResNet50",with_fpn=True)

model.train(
    num_epochs=12,
    train_dataset=train_dataset,
    train_batch_size=2,
    eval_dataset=eval_dataset,
    learning_rate=0.0025,
    lr_decay_epochs=[8,11],
    warmup_steps=100,
    warmup_start_lr=0.00025,
    save_dir="output/faster_rcnn_fpn-bottle",
    use_vdl=True
)
```

```python
#创建本地简单的AI代理，本地大模型WebAPI接口
import requests
import json
# Ollama 服务的 URL，假设它运行在本地的 11434 端口
base_url = 'http://localhost:11434/api/generate'
model_name = 'qwen2.5'
# 发送请求的函数
def generate_completion_stream(prompt):
    # 构造请求的 JSON 数据,流式输出
    data = {
        "model": model_name,
        "prompt": prompt,
        "max_tokens": 1024,
        "temperature": 0.7,
        "top_p": 0.9,
        "stream": True
    }
    # 发送 POST 请求
    response = requests.post(base_url, json=data,stream=True)
    #流式生成文本补全
    result=""
    for line in response.iter_lines():
        if line:
            linedict=json.loads(line.decode('utf-8'))
            print(linedict["response"],end="")
            result+=linedict["response"]
    return result


# 测试发送请求
prompt = "如何检测AI生成的文本"
stream_completion= generate_completion_stream(prompt)
```

```python
# 大模型文本生成
# PaddleNLP 提供了方便易用的 Auto API，能够快速的加载模型和 Tokenizer。这里以使用 Qwen/Qwen2-0.5B 模型做文本生成为例：

from paddlenlp.transformers import AutoTokenizer, AutoModelForCausalLM
tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2-0.5B")
model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2-0.5B", dtype="float16")
input_features = tokenizer("你好！请自我介绍一下。", return_tensors="pd")
outputs = model.generate(**input_features, max_length=128)
print(tokenizer.batch_decode(outputs[0], skip_special_tokens=True))

# ['我是一个AI语言模型，我可以回答各种问题，包括但不限于：天气、新闻、历史、文化、科学、教育、娱乐等。请问您有什么需要了解的吗？']
```
